```python
import pandas as pd
```

```python
df=pd.read_csv("financial_dataset.csv")
```

```python
df
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df["Sales"]=df["Sales"].fillna(df["Sales"].mean())
```

```python
df["Profit"]=df["Profit"].fillna(df["Profit"].mean())
```

```python
df["Category"]=df["Category"].fillna(df["Category"].mode())
```

```python
df["City"]=df["City"].fillna(df["City"].mode())
```

```python
df.duplicated().sum()
```

```python
df["Category"]=df["Category"].str.strip().str.title()
```

```python
df["Payment_Status"]=df["Payment_Status"].str.strip().str.title()
```

```python
df["City"]=df["City"].str.strip().str.title()
```

```python
df["Transaction_Date"]=pd.to_datetime(df["Transaction_Date"],errors="coerce")
df.dtypes                                     
```

```python
print("Cleaned Financial Dataset:")
df
```
