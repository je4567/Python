```python
numeric_columns = df.select_dtypes(include="number").columns

print("\nNumerical Columns:")
print(numeric_columns.tolist())
```

```python
if len(numeric_columns) > 0:

    for column in numeric_columns:

        print("\n--------------------------------")
        print("Analysis for:", column)
        print("--------------------------------")

        print("Minimum:", round(df[column].min(), 2))
        print("Maximum:", round(df[column].max(), 2))
        print("Average:", round(df[column].mean(), 2))
        print("Current:", round(df[column].iloc[-1], 2))

```

```python
plt.figure(figsize=(10, 6))
plt.violinplot(
    [
        df["Open"],
        df["High"],
        df["Low"],
        df["Close"],
        df["Adj Close"]
    ],
    showmeans=True,
    showmedians=True
)
plt.xticks(
    [1, 2, 3, 4, 5],
    ["Open", "High", "Low", "Close", "Adj Close"]
)
plt.title("Walmart Stock Price Distribution - Violin Plot")
plt.xlabel("Stock Price Variables")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.pairplot(df,
            x_vars=["Date","Open","Close"],
            y_vars=["High","Low"],
            diag_kind="kde")
plt.show()
```

```python
correalation=df[['Open','High','Low','Close','Adj Close','Volume']].corr()
print(correalation)
```

```python
# --------------------------------------------------
# FINANCIAL DASHBOARD SUMMARY
# --------------------------------------------------

print("\n======================================")
print("       FINANCIAL DASHBOARD SUMMARY")
print("======================================")

print("Total Records:", len(df))

print("Date Range:",
      df["Date"].min().strftime("%d-%m-%Y"),
      "to",
      df["Date"].max().strftime("%d-%m-%Y"))

print("Current Price:", round(df["Close"].iloc[-1], 2))

print("Highest Price:", round(df["Close"].max(), 2))

print("Lowest Price:", round(df["Close"].min(), 2))

print("Average Closing Price:", round(df["Close"].mean(), 2))

print("Average Daily Return:",
      round(df["Daily Return"].mean(), 2), "%")

print("Volatility:",
      round(df["Daily Return"].std(), 2), "%")

print("Highest Trading Volume:",
      int(df["Volume"].max()))

print("Average Trading Volume:",
      round(df["Volume"].mean(), 0))


# --------------------------------------------------
# GROUPBY SUMMARY BY YEAR
# --------------------------------------------------
yearly_summary = df.groupby("Year").agg(
    Number_of_Records=("Close", "count"),
    Average_Close=("Close", "mean"),
    Highest_Close=("Close", "max"),
    Lowest_Close=("Close", "min"),
    Average_Volume=("Volume", "mean"),
    Total_Volume=("Volume", "sum"),
    Average_Daily_Return=("Daily Return", "mean"),
    Volatility=("Daily Return", "std")
).round(2)


# --------------------------------------------------
# DISPLAY YEARLY GROUPBY SUMMARY
# --------------------------------------------------

print("\n======================================")
print("          YEARLY GROUPBY SUMMARY")
print("======================================")

print(yearly_summary)


# --------------------------------------------------
# GROUPBY SUMMARY BY DATE
# --------------------------------------------------

data_summary = df.groupby("Date").agg(
    Number_of_Records=("Close", "count"),
    Average_Close=("Close", "mean"),
    Highest_Close=("Close", "max"),
    Lowest_Close=("Close", "min"),
    Total_Volume=("Volume", "sum")
).sort_values("Total_Volume", ascending=False)


print("\n======================================")
print("           DAILY GROUPBY SUMMARY")
print("======================================")

print(data_summary.head(20))


# --------------------------------------------------
# YEARLY AVERAGE CLOSING PRICE
# --------------------------------------------------

yearly_close = df.groupby("Year")["Close"].mean()


print("\n======================================")
print("       YEARLY AVERAGE CLOSE")
print("======================================")

print(yearly_close.round(2))


# --------------------------------------------------
# KEY FINANCIAL INSIGHTS
# --------------------------------------------------

highest_year = yearly_close.idxmax()
lowest_year = yearly_close.idxmin()

overall_change = (
    (df["Close"].iloc[-1] - df["Close"].iloc[0])
    / df["Close"].iloc[0]
) * 100

print("\n======================================")
print("          KEY FINANCIAL INSIGHTS")
print("======================================")

print(
    "Year with Highest Average Closing Price:",
    highest_year
)

print(
    "Year with Lowest Average Closing Price:",
    lowest_year
)

print(
    "Overall Price Change:",
    round(overall_change, 2),
    "%"
)

print(
    "Average Daily Return:",
    round(df["Daily Return"].mean(), 2),
    "%"
)

print(
    "Daily Volatility:",
    round(df["Daily Return"].std(), 2),
    "%"
)
```
