```python
df["Date"]=pd.to_datetime(df["Date"])
df.dtypes
```

```python
df.isna().sum()
```

```python
df['Close']=df['Close'].ffill()
```

```python
df['Open']=df['Open'].bfill()
```

```python
df['Volume']=df['Volume'].ffill()
```

```python
df['High']=df['High'].ffill()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df=df.sort_values("Date")
df
```

```python
df["MA_5"]=df["Close"].rolling(5).mean() 
df["MA_20"]=df["Close"].rolling(20).mean()
df
```

```python
plt.figure(figsize=(12,6))
plt.plot(df['Date'],df['Close'],label='Actual Close Price',color='blue',marker='o')
plt.plot(df['Date'],df['MA_5'],label='5-Day Moving Average',color='red')
plt.plot(df['Date'],df['MA_20'],label='20-Day Moving Average',color='pink')
plt.title('Stock Price Trend & Rolling MOving Averages')
plt.xlabel('Timeline')
plt.ylabel('Stock Value')
plt.grid(True)
plt.legend()
plt.xticks(rotation=30)
plt.show()
```
