```python
try:
    balance=int(input("enter the balance"))
    T=int(input("enter the transactions"))
    transactions=[]
    for i in range(T):
        print("\n Transactions",i+1)
        try:
            amount=int(input("enter the amount:"))
            transaction_type=input("Debit or Credit:")
            transactions.append([amount,transaction_type])
        except ValueError:
            print("invalid input,enter the valied input")

    for transaction in transactions:
        amount=transaction[0]
        transaction_type=transaction[1]

    if transaction_type.lower()=="credit":
        balance=balance+amount
    elif transaction_type.upper()=="debit":
        balance=balance-amount
    else:
        print("invailed input")
        
    print("Final balance",balance)
except ValueError:
    print("invalid input,enter the valied input")
```
