```python
sales_report=df.groupby("Category").agg(
    Total_sales=("Sales","sum"),
    Total_profit=("Profit","sum"),
    Avg_sales=("Sales","mean"),
    Total_quantity=("Quantity","sum")
)
print(sales_report)
```

```python
pivot=pd.pivot_table(
    df,
    values="Sales",
    index="Customer",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)
print(pivot)
    
```
