```python
df["current_ratio"]=(df["Current_Assets"]/(df["Current_Liabilities"]))
```

```python
df["ROE"]=(df["Net_Income"]/(df["Equity"])*100)
```

```python
df["EPS"]=(df["Net_Income"]/(df["Shares_Outstanding"]))
```

```python
df
```

```python
Max_current_ratio=df['current_ratio'].max()
print("Max ratio:",Max_current_ratio)
```

```python
Max_EPS=df['EPS'].max()
print("Max ratio:",Max_EPS)
```

```python
Max_ROE=df['ROE'].max()
print("Max ratio:",Max_ROE)
```

```python
print("Financial Ratio Analysis")
print("_________")
df[["Year","Net_Income","current_ratio","ROE","EPS"]]
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Year"],df["current_ratio"])
plt.title("current_ratio trend")
plt.xlabel("year")
plt.ylabel("current ratio")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Year"],df["EPS"],marker="o")
plt.title("EPS trend")
plt.xlabel("year")
plt.ylabel("EPS")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Year"],df["ROE"],marker="o")
plt.title("ROE trend")
plt.xlabel("year")
plt.ylabel("ROE")
plt.grid(True)
plt.show()
```
