```python
total_sales=df["Sales"].sum()
total_expenses=df["Expenses"].sum()
total_Profit=df["Profit"].sum()
total_Assets=df["Assets"].sum()
total_Liabilities=df["Liabilities"].sum()
print("total sales:",total_sales)
print("total expenses:",total_expenses)
print("total Profit:",total_Profit)
print("total Assets:",total_Assets)
print("total Liabilities:",total_Liabilities)
```

```python
average_sales=df["Sales"].mean()
highest_sales=df["Sales"].mean()
average_stock_price=df["Stock_Price"].mean()
print("average sales:",average_sales)
print("highest sales:",highest_sales)
print("average stock price :",average_stock_price)
```

```python
monthly_sales=df.groupby("Month")[["Sales","Expenses","Profit"]].mean()
print("            Monthly Financial Summary:-")
print(monthly_sales)
```

```python
yearly_summary=df.groupby("Year")[["Sales","Expenses","Profit"]].mean()
print("            Yearly Financial Summary:-")
print(yearly_summary)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))
plt.plot(
    monthly_sales.index,
    monthly_sales["Profit"],
    marker="o",
    alpha=0.7,
    label="Profit")
plt.plot(
    monthly_sales.index,
    monthly_sales["Sales"],
    marker="o",
    alpha=0.7,
    label="Sales",
)
plt.title("monthly sales")
plt.xlabel("sales")
plt.ylabel("profit")
plt.legend()
plt.grid(True)
plt.xticks(rotation=40)
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(
    yearly_summary.index,
    yearly_summary["Sales"],
    marker="o",
    alpha=0.7,
    label="sales")
plt.title("monthly sales")
plt.xlabel("sales")
plt.ylabel("year")
plt.legend()
plt.grid(True)
plt.xticks(rotation=40)
plt.show()
```

```python
plt.figure(figsize=(16,8))
sns.barplot(x="Month",y="Sales",data=df,width=0.5)
plt.title("category vs profit")
plt.xlabel("month")
plt.ylabel("sales")
plt.title("monthly sales")
plt.xticks(rotation=40)
plt.show()
```

```python
profit_margin=((df["Profit"])/df["Sales"])*100
print("profit margin:",profit_margin)
```

```python
plt.figure(figsize=(16,8))
sns.scatterplot(x="Sales",y="Profit",data=df,hue="Sales")
plt.title("sales vs profit")
plt.xlabel("sales")
plt.ylabel("month")
plt.grid(True)
plt.xticks(rotation=40)
plt.show()
```

```python
numeric_cols=["Sales","Expenses","Profit","Assets","Liabilities","Stock_Price"]
correlation=df[numeric_cols].corr()
correlation
```

```python
sns.heatmap(correlation,
            annot=True,
            cmap='Blues',
            cbar=False,
            square=True,
            linewidth=1)
plt.show()
```
