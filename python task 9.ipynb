```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Marks.csv")
df
```

```python
numeric_cols=["Math","Science","English","Computer","Attendance"]
jenevia=df[numeric_cols].corr()
jenevia
```

```python
sns.heatmap(jenevia,
            annot=True,
            cmap="Reds",
            cbar=False,
            square=True,
            linewidth=1.5)
plt.title("Marks")
plt.show()

            
```

```python
sns.barplot(x="Math",y="Science",data=df,width=1,color="blue")
plt.title("math vs science")
plt.show()
```

```python
sns.lineplot(x="Math",y="English",data=df)
plt.title("math vs English")
plt.show()
```

```python
data={
    "department":["CA","science","HR","physical_education","arts","electronc","AI","accounts"],
    "budget":[54000,25600,25478,62000,54700,23600,58800,41000],
     "actual":[55000,26000,25000,60000,50000,25000,60000,40000]
}
df=pd.DataFrame(data)
df
```

```python
df_long=df.melt(
    id_vars="department",
    value_vars=["budget","actual"],
    var_name="type",
    value_name="amount"
)
df_long

sns.barplot(
    data=df_long,
    x="department",
    y="amount",
    hue="type"
)   

plt.title("budget VS actual")
plt.xlabel("Deparment")
plt.ylabel("Amount")
plt.show()
```
